# Caderno 7.1 - Pesquisa com BM25 [runs]

## 1. Parâmetros

```python
PASTA_DOCS_QRELS = './dados/outputs/0 - qrel - docs - query - raw_human_eval/'
PASTA_CADERNO_INDICES = './dados/outputs/5 - indexacao bm25/'

PASTA_RESULTADO_CADERNO = './dados/outputs/7 - runs/'

ARQUIVO_QRELS = f'{PASTA_DOCS_QRELS}/qrel.csv'
ARQUIVO_QUERIES = f'{PASTA_DOCS_QRELS}/query.csv'

NOME_ARQUIVO_INDICE_TEXTO = f'{PASTA_CADERNO_INDICES}indice_bm25_texto.pickle'
NOME_ARQUIVO_INDICE_TEXTO_E_ASSUNTO = f'{PASTA_CADERNO_INDICES}indice_bm25_texto_e_assunto.pickle'

RUNS = [
    {
        "arquivo": "run_bm25_0.9_0.4_assunto_e_texto.csv",
        "k1": 0.9,
        "b": 0.4,
        "nome_indice": "ASSUNTO+TEXTO"
    },
    {
        "arquivo": "run_bm25_0.9_0.4_texto.csv",
        "k1": 0.9,
        "b": 0.4,
        "nome_indice": "TEXTO"
    },
    {
        "arquivo": "run_bm25_0.82_0.68_assunto_e_texto.csv",
        "k1": 0.82,
        "b": 0.68,
        "nome_indice": "ASSUNTO+TEXTO"
    },
    {
        "arquivo": "run_bm25_0.82_0.68_texto.csv",
        "k1": 0.82,
        "b": 0.68,
        "nome_indice": "TEXTO"
    },
    {
        "arquivo": "run_bm25_1_0.3_assunto_e_texto.csv",
        "k1": 1,
        "b": 0.3,
        "nome_indice": "ASSUNTO+TEXTO"
    },
    {
        "arquivo": "run_bm25_1_0.3_texto.csv",
        "k1": 1,
        "b": 0.3,
        "nome_indice": "TEXTO"
    }
]
```

## 2. Carrega as queries e o qrels

```python
import pandas as pd

queries = pd.read_csv(ARQUIVO_QUERIES)
qrels = pd.read_csv(ARQUIVO_QRELS)
```

## 3. Carrega os 2 índices: apenas o texto da norma / apenas o texto da norma e assunto

```python
from bm25 import IndiceInvertido, BM25, tokenizador_pt_remove_html

import os

# Índice invertido para o texto e para o texto+assunto
iidx_texto = IndiceInvertido(tokenizador_pt_remove_html)
iidx_texto_e_assunto = IndiceInvertido(tokenizador_pt_remove_html)

iidx_texto.from_pickle(NOME_ARQUIVO_INDICE_TEXTO)
iidx_texto_e_assunto.from_pickle(NOME_ARQUIVO_INDICE_TEXTO_E_ASSUNTO)
```

## 4. Executa todos os experimentos

```python
from metricas import metricas

for run in RUNS:
    arquivo = run['arquivo']
    k1 = run['k1']
    b = run['b']
    indice = iidx_texto if run['nome_indice'] == 'TEXTO' else iidx_texto_e_assunto

    print(f'Executando experimento: {arquivo}')

    buscador = BM25(indice, k1, b, bias_idf=1)

    col_resultado_query_key=[]
    col_resultado_doc_key=[]
    col_resultado_rank=[]
    
    for i, row in queries.iterrows():
        query_key = row.KEY
        query_text = row.TEXT
        resultados = buscador.pesquisar(query_text)

        # Salva os 1000 primeiros documentos da busca
        n_docs = 1000
        primeiros_docs = [tupla_key_score[0] for tupla_key_score in resultados[:1000]]
        queries_keys = [query_key] * len(primeiros_docs)
        ranking = list(range(1, len(primeiros_docs)+1))
    
        col_resultado_query_key.extend(queries_keys)
        col_resultado_doc_key.extend(primeiros_docs)
        col_resultado_rank.extend(ranking)
    
    df_resultados = pd.DataFrame({
        "QUERY_KEY": col_resultado_query_key,
        "DOC_KEY": col_resultado_doc_key,
        "RANK": col_resultado_rank,
    })
    df_resultados.to_csv(f'{PASTA_RESULTADO_CADERNO}{arquivo}', index=False)

    df_metricas = metricas(df_resultados, qrels, aproximacao_trec_eval=True, k=[10])
    print(f'\tnDCG@10: {df_metricas['nDCG@10'].mean():.4f}')
```
