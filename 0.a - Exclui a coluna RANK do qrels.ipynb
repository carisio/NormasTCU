# Caderno 0.a - Exclui a coluna RANK do qrels

Todos os qrels foi gerado com uma coluna RANK, mas não precisa delas.

```python
import os
import pandas as pd

# Caminho base
base_path = "./dados/outputs/0 - qrel - docs - query - raw_human_eval/"

# Percorrer diretórios e subdiretórios
for root, dirs, files in os.walk(base_path):
    for file in files:
        # Verifica se o nome começa com 'qrel' e termina com '.csv'
        if file.startswith("qrel") and file.endswith(".csv"):
            file_path = os.path.join(root, file)
            
            try:
                # Ler o CSV
                df = pd.read_csv(file_path)
                
                # Remover a coluna 'RANK' se existir
                if "RANK" in df.columns:
                    df = df.drop(columns=["RANK"])
                    
                    # Salvar sobrescrevendo o arquivo original
                    df.to_csv(file_path, index=False)
                    print(f"Atualizado: {file_path}")
                else:
                    print(f"Sem coluna RANK: {file_path}")
            
            except Exception as e:
                print(f"Erro ao processar {file_path}: {e}")
```
